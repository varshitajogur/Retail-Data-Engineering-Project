# 01 — Ingestion and Schema Validation

Loads the repository datasets and validates their schemas before transformation.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
spark = SparkSession.builder.appName('RetailIngestion').getOrCreate()
```

```python
base = '../Datasets/'
customers = spark.read.option('header', True).option('inferSchema', True).csv(base+'customers.csv')
orders = spark.read.option('header', True).option('inferSchema', True).csv(base+'orders_new.csv')
order_items = spark.read.option('header', True).option('inferSchema', True).csv(base+'order_items.csv')
```

```python
customers.printSchema()
orders.printSchema()
order_items.printSchema()
```

```python
print('customers:', customers.count())
print('orders:', orders.count())
print('order_items:', order_items.count())
```
